## Project Context

### Scenario

Der Telekommunikationsanbieter Teleconfia versucht, in den USA Fuss zu fassen. Als erstes Probegebiet wurde Florida gewaehlt. Neue Kunden konnten ein Jahr lang zu guenstigen Konditionen das Teleconfia-Netz nutzen. Nicht alle Kunden sind geblieben — diese Abwanderung (Churn) soll durch zwei Marketingkampagnen bekaempft werden.

### Mission

1. **Plakatkampagne** — die vier Stadtgebiete mit der hoechsten Kundenabwanderung identifizieren.
2. **Individualkontakt** — Kunden identifizieren, die wahrscheinlich abwandern, um ihnen Sonderkonditionen anzubieten.

Dafuer muessen geeignete Indikator-Variablen (kategorisch, Integer, Float) gefunden und Grenzwerte fuer die Kontaktaufnahme bestimmt werden — der Float-Grenzwert wird per logistischer Regression hergeleitet.

### Scope / Hypotheses

* Es gibt Stadtgebiete mit deutlich ueberdurchschnittlicher Abwanderung.
* Bestimmte Vertragsmerkmale (z.B. `international_plan`) korrelieren mit Churn.
* Ab einer gewissen Anzahl `customer_service_calls` steigt das Abwanderungsrisiko sprunghaft.
* Nutzungsintensitaet (`total_day_minutes`) besitzt einen Schwellenwert, ab dem Churn wahrscheinlicher als Verbleib wird.

### Methode & Metrics

| Metrik | Zielwert | Begruendung |
|--------|----------|-------------|
| Churn-Rate gesamt | Baseline erheben | Ausgangspunkt fuer alle Vergleiche |
| Top-4 Staedte nach Churn-Anzahl | Auswahl fuer Plakatkampagne | Absolute Abwanderungszahl bestimmt Kampagnen-Impact vor Ort |
| Grenzwert `customer_service_calls` | ab 3-4 Anrufen kontaktieren | Fruehwarnung vor dem Punkt, an dem Churn-Rate stark ansteigt (ab 5) |
| Grenzwert `total_day_minutes` | via Logistic Regression (50%-Schwelle) | Objektiver statistischer Schwellenwert statt Bauchgefuehl |

### Conventions

| Variable | Phase | Zustand |
|----------|-------|---------|
| `df_raw` | Loading | Originalzustand – schreibgeschützt |
| `df_eda` | Exploration | Initiale Daten zur explorativen Analyse |
| `df_edit` | Preparation & Processing | Bereinigt, transformiert & aggregiert (Arbeitsbasis) |
| `df_final` | Analysis & Reporting | Finaler Output für Insights und Visualisierungen |

> **Best Practice:** Nutze für jeden Transformationsschritt `.copy()`, um die Traceability zu gewährleisten und `df_raw` niemals zu überschreiben.
